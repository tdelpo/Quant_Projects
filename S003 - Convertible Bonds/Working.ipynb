## Valuation of Convertible Bond

```python
import math
from scipy.stats import norm

# Bond Parameters
face_value = 1000
bond_maturity = 15
yeild = 0.05
coupon_rate = 0.06
spread = 0.02
discount_rate = yeild + spread

# Stock parameters
conversion_ratio = 30
conversion_price = 25

# Black-Scholes parameters
S = 25 #Current Stock Price 
sigma = 0.2 #Volatility
r_f = 0.03 #Risk-free rate
K = 25 #Strike Price
T = 10 #Maturity

# Present Value Cash Flow
year = [x for x in range(1,T+1)]
coupon_cash = [face_value*coupon_rate for _ in year]
principle_cash = [face_value if x == T else 0 for x in year] 
total_cash = [coupon_cash[i] + principle_cash[i] for i in range(len(year))]
discounted_cash = [round(total_cash[i]/(1+discount_rate)**year[i],3) for i in range(len(year))]
discounted_cash_value = sum(discounted_cash)
print(f"Present Value of Total Cash Flow: £{discounted_cash_value}")

# Call option value
d1 = (math.log(S/K) + (r_f + (sigma**2)/2)*T) / (sigma*math.sqrt(T))
d2 = d1 - (sigma*math.sqrt(T))
call_option_price = (norm.cdf(d1)*S) - (norm.cdf(d2)*K*math.exp(-r_f*T))
total_call_option_value = call_option_price * conversion_ratio
print(f"Call Option Pricing:\n",
      f"d1 = {round(d1,2)}\n", 
      f"d2 = {round(d2,2)}\n", 
      f"Price = £{round(call_option_price,2)}\n", 
      f"Total Value: £{round(total_call_option_value,2)}.")

# Total Convertible Bond Value
convertible_bond_value = discounted_cash_value + total_call_option_value 
print(f"Convertible Bond Value: £{round(convertible_bond_value,2)}")
```
