### Bubble Sort Function

Write the bubble sort function to sort the list of numbers and describe how its working

```python
# Description

# The bubble sort function works by carrying the largest number is finds to the last spot on in the list. Each pass of the list sorts the list by biggest numbers first (to the rear) and to the front. 
# Time complexity = O(n^2)

# Function

def bubble_sort(my_list):
    for i in range(len(my_list)-1, 0, -1):
        for j in range(i):
            if my_list[j] > my_list[j+1]:
                my_list[j], my_list[j+1] = my_list[j+1], my_list[j]
    return my_list

# Run Test
for test in test_cases:
    print(bubble_sort(test))
```

### Selection Sort Function

Write the selection sort function to sort the list of numbers

```python
# Description

# Working from the front of the list to the back, in each loop the selection sort finds the smallest number in the array and places it into the indexed position. This is the opposite to the bubble sort in that smallest numbers will start to be arranged first rather than the largest numbers.
# Time complexity = O(n^2)

# Function
def selection_sort(my_list):

    for i in range(len(my_list) - 1):
        min_index = i
        for j in range(i+1, len(my_list)):
            if my_list[j] < my_list[min_index]:
                min_index = j
        if i != min_index:
            my_list[min_index], my_list[i] = my_list[i], my_list[min_index]

    return my_list

# Run Test
for test in test_cases:
    print(selection_sort(test))
```

```python
# Answer
def selection_sort(my_list):    
    for i in range(len(my_list)-1):
        min_index = i 
        for j in range(i+1, len(my_list)):
            if my_list[j] < my_list[min_index]:
                min_index = j
        if i != min_index:
            my_list[i], my_list[min_index] = my_list[min_index], my_list[i] 
    return my_list
```

### Insertion Sort Function

Write the insertion sort function to sort the list of numbers

```python
# Description

# The insertion sort algorithm moves progressively forward through a list and back sorts the values behind it. That is, it indexes and new value each loop and finds its place in the values which it has previously visited. Unlike bubble and selection sorts, it isn't biased toward sorting the minimum or maximum values first as it has no guarantee that it's seen said values and sorted them accordingly. 

# Function
def insertion_sort(my_list):

    for i in range(1, len(my_list)):
        temp = my_list[i]
        j = i-1
        while my_list[j] > temp and j > -1:
            my_list[j+1] = my_list[j]
            my_list[j] = temp
            j -= 1

    return my_list

# Run Test
for test in test_cases:
    print(insertion_sort(test))
```

### Merge Sort Function

Write the merge sort function to sort the list of numbers

```python
# Description

# The merge sort function works by halving an array into left and right until there are singular elements. The singular elements are then merged back together in order using pointers to create the combined array. 
# Time complexity: O(n log n)

# Function
def merge(list1, list2):
    combined = []
    i, j = 0, 0
    while i < len(list1) and j < len(list2):
        if list1[i] < list2[j]:
            combined.append(list1[i])
            i += 1
        else:
            combined.append(list2[j])
            j += 1
    while i < len(list1):
        combined.append(list1[i])
        i += 1
    while j < len(list2):
        combined.append(list2[j])
        j += 1
    return combined

def merge_sort(my_list):

    if len(my_list) == 1:
        return my_list
    mid_index = int(len(my_list)/2)
    left = merge_sort(my_list[:mid_index])
    right = merge_sort(my_list[mid_index:])
    return merge(left, right)

# Run Test
for test in test_cases:
    print(merge_sort(test))
```

### Quick Sort Function

Write the quick sort function to sort the list of numbers

```python
# Description

# The quick sort algorithm is recursive. It starts by selecting a pivot index within a list and sorting or numbers less than the pivot index value to the left, and numbers greater to the right. Once the halves are split, it recursively sorts each half according to the same logic. The algorithm recursively calls until lists are of lengths zero and/or one and returns the results. 
# Time complexity = O(n log n)

# Function
def swap(my_list, index1, index2):
    temp = my_list[index1]
    my_list[index1] = my_list[index2]
    my_list[index2] = temp 
    return my_list

def pivot(my_list, pivot_index, end_index):
    swap_index = pivot_index
    for i in range(pivot_index+1, end_index+1):
        if my_list[i] < my_list[pivot_index]:
            swap_index += 1
            swap(my_list, swap_index, i)
    swap(my_list, pivot_index, swap_index)
    return swap_index

def quick_sort_helper(my_list, left, right):
    if left < right:
        pivot_index = pivot(my_list, left, right)
        quick_sort_helper(my_list, left, pivot_index-1)
        quick_sort_helper(my_list, pivot_index+1, right)
    return my_list

def quick_sort(my_list):
    return quick_sort_helper(my_list, 0, len(my_list)-1)

# Run Test
for test in test_cases:
    print(quick_sort(test))
```

```python
# Answer
def swap(my_list, index1, index2):
    temp = my_list[index1] 
    my_list[index1] = my_list[index2] 
    my_list[index2] = temp  

def pivot(my_list, pivot_index, end_index):
    swap_index = pivot_index 
    for i in range(pivot_index+1, end_index+1): 
        if my_list[i] < my_list[pivot_index]: 
            swap_index += 1
            swap(my_list, swap_index, i) 
    swap(my_list, pivot_index, swap_index) 
    return swap_index

def quick_sort_helper(my_list, left, right):
    if left < right:
        pivot_index = pivot(my_list, left, right)
        quick_sort_helper(my_list, left, pivot_index-1) 
        quick_sort_helper(my_list, pivot_index+1, right) 
    return my_list 

def quick_sort(my_list): 
    return quick_sort_helper(my_list, 0, len(my_list)-1)
```
