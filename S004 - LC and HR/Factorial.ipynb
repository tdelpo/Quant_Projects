Write a function to return the factorial of a given input integer

```python
# Recursive
def factorial(n):
    if n == 1:
        return 1
    return n* factorial(n-1)

def factorial_tab(n):
    dp = [0]*(n+1) 
    for i in range(1, n+1):
        if i == 1:
            dp[i] = 1
        else:
            dp[i] = i * dp[i-1]
    return dp[-1]
```

```python
# Test cases
tests = [1, 2, 3, 4]

# Run
for test in tests:
    print(factorial(test))

# Alt. Run
for test in tests:
    print(factorial_tab(test))
```
