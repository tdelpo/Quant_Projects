A positive integer is 'fancy' if its represetation in base 4 only includes 0s and 1s. For example: 17 is fancy as its base-4 representation is 101, but 18 is not as 102 includes a 2. 

Given a psitive integer n, find the number of fancy positive integers less than n.

Note that n may be very large and the algorithm should be faster than iterating over all values less than n. 

Note:
n = 1, output is 0 since no positive integers less than 1. 
n = 10, output is 3, since 1, 4, 5 are fancy

```python
# Function

def FancyBase4(n):

    # Validate input
    if not isinstance(n, int):
        raise TypeError("n must be an integer") 

    # Edge case 1
    if n <= 1:
        return 0

    # Base 4 of a n
    nums = []
    i = n 
    while i > 0:
        nums.append(i % 4)
        i //= 4
    nums.reverse() 
    
    # Number of nums:
    L = len(nums) 

    # Fancy numbers less than L = 2^(j-1)
    count = sum(1 << (j-1) for j in range(1, L))

    # Find fancy nums with same length
    for i, j in enumerate(nums):

        left = L - i - 1

        if i == 0:
            permit = [1] # stop first '0' 
        else:
            permit = [0,1] 
        
        lessThan = sum(p < j for p in permit) 
        count += lessThan * (1 << left) 

        # Catch non-fancy n in nums:
        if j not in permit:
            break

    return count
```

```python
# Test Cases
tests = [1, 10]

# Run 
for test in tests:
    print(FancyBase4(test))
```
