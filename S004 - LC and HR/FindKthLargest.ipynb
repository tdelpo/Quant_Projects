Given a sequence of n numbers, how would you efficiently find the kth largest number?

```python
import random

def kth_largest(arr, k):
    k = len(arr) - k      # convert to kth smallest index

    left, right = 0, len(arr) - 1

    while True:
        pivot = random.randint(left, right)
        arr[pivot], arr[right] = arr[right], arr[pivot]

        # Lomuto partition
        store = left
        for i in range(left, right):
            if arr[i] < arr[right]:
                arr[i], arr[store] = arr[store], arr[i]
                store += 1

        arr[store], arr[right] = arr[right], arr[store]

        if store == k:
            return arr[store]
        elif store < k:
            left = store + 1
        else:
            right = store - 1
```

What an interviewer is often looking for

There are two answers depending on the emphasis:

Quickselect: expected O(n), simple and widely used.

Median of Medians: deterministic O(n) worst-case by choosing a guaranteed good pivot, though it's more complex and less common in production.

For most interviews, I'd lead with Quickselect and mention that the worst case is O(n²), adding that median-of-medians removes that worst case if required
