Write a function to return the nth fib number in the fib series

```python
# Function
def fib(n):

    if not isinstance(n, int) or n < 1:
        raise TypeError("n must be integer greater than 0")

    if n == 1 or n == 2:
        return 1

    return fib(n-1) + fib(n-2)
```

```python
# Test cases
tests = [
    1,
    2,
    3,
    4,
    5
]

# Run
for test in tests:
    print(fib(test))
```
