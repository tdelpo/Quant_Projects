```python
# Helper Function
def merge_row_left(row):

    # Remove the zeroes
    digits = [x for x in row if x != "."] 

    # Merge same numbers
    new_row = []
    i = 0
    while i < len(digits):
        if i + 1 < len(digits) and digits[i] == digits[i+1]:
            new_row.append(digits[i]*2) 
            i += 2
        else:
            new_row.append(digits[i]) 
            i += 1

    # Fill with zeros
    new_row = new_row + ["."] * (4 - len(new_row)) 

    return new_row 

def solution(moves):

    # Create board
    move = move
    n = 4
    m = 4
    board = [["."]*m for _ in range(n)]

    # Moves
    for move in moves:

        # Assign variables
        move = move.split(" ") 
        DIR = move[0]
        j,i = int(move[1]), int(move[2])
        value = int(move[3]) 

        # Consolidate board based on move direction
        if DIR == "L": # merge left
            board = [merge_row_left(row) for row in board] 

        if DIR == "R":
            board = [list(reversed(merge_row_left(list(reversed(row))))) for row in board]

        if DIR == "U":
            trans = [list(col) for col in zip(*board)] #unpack, transport, make list
            trans = [merge_row_left(row) for row in trans]
            board = [list(row) for row in zip(*trans)] #reverse trans method

        if DIR == "D":
            trans = [list(col) for col in zip(*board)] #unpack, transport, make list
            trans = [list(reversed(merge_row_left(list(reversed(row))))) for row in trans]
            board = [list(row) for row in zip(*trans)] #reverse trans method

        # Insert new value to board
        if board[i][j] != ".":
            return "Error" 
        else:
            board[i][j] = value 

    # Convert board to output format 
    board = "\n".join(",".join(str(x) for x in row) for row in board)

    return board
```
