Write a function to return prime numbers less than or equal to a given integer with the option to return prime numbers ending in a given integer.

```python
# Function - 1

def PrimeNumbers(n, endsWith = None):
    """ Returns list of prime numbers less than or equal to n.""" 

    nums = [1] * (n+1) 
    nums[0] = 0
    
    for i in range(2, n+1):
        if nums[i] == 1:
            for j in range(i*2, n, i):
                nums[j] = 0

    primes = []

    for idx, num in enumerate(nums):
        if num == 1:
            primes.append(idx)

    if endsWith is not None:
        prime_endsWith = []
        for num in primes:
            if num%10 == endsWith:
                prime_endsWith.append(num)
        
    return primes if endsWith is None else prime_endsWith
```

```python
# Test Cases
tests = [30]

# Run
for test in tests:
    print(PrimeNumbers(test))

# Run alt
for test in tests:
    print(PrimeNumbers(test, 3))
```
