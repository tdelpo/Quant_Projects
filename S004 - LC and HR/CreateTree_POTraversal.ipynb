Given a list of lists where each list contains the key, value and children of a tree. Write a function to create the tree and return the pre-order traversal value list.

```python
# Node class function
class Node:

    def __init__(self, key, value):
        self.key = key 
        self.value = value 
        self.children = [] 

    def __repr__(self):
        return f"Node: ({self.key}, {self.value}, children={self.children})"


# Traversal Function
def preorder(root):

    result = []

    def traverse(node):
        result.append(node.key)
        result.append(node.value)

        for child in node.children:
            traverse(child)

    traverse(root)

    return result

# Main Function
def solution(arg1):

    # Create all nodes
    nodes = [Node(row[0], row[1]) for row in arg1]

    # Attach children
    hasParent = [False] * len(arg1) 
    for i, row in enumerate(arg1):
        for c_i in row[2:]:
            nodes[i].children.append(nodes[c_i]) 
            hasParent[c_i] = True 
    
    # Find the root
    roots = [nodes[i] for i in range(len(nodes)) if not hasParent[i]]
    root = roots[0]

    # Pre-order traversal
    result = preorder(root)

    return result
```

```python
# Test Cases
arg1 = [
    [4, 2], 
    [1, 15, 3, 2, 0], 
    [1, 0, 4],
    [6, 3],
    [2, 3]]

# Run
print(solution(arg1))
```
