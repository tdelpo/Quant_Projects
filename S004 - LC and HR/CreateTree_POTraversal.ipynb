Given a list of lists where each list contains the key, value and children of a tree. Write a function to create the tree and return the pre-order traversal value list.

```python
# Node class function
class Node:

    def __init__(self, key, value):
        self.key = key 
        self.value = value 
        self.children = [] 

    def __repr__(self):
        return f"Node: ({self.key}, {self.value}, children={self.children})"


# Traversal Function
def preorder(root):

    result = []

    def traverse(node):
        result.append(node.key)
        result.append(node.value)

        for child in node.children:
            traverse(child)

    traverse(root)

    return result

# Create Tree Function
def create_tree(argX):

    # Create all nodes
    nodes = [Node(row[0], row[1]) for row in arg1]

    # Attach children
    hasParent = [False] * len(arg1) 
    for i, row in enumerate(arg1):
        for c_i in row[2:]:
            nodes[i].children.append(nodes[c_i]) 
            hasParent[c_i] = True 
    
    # Find the root
    roots = [nodes[i] for i in range(len(nodes)) if not hasParent[i]]
    root = roots[0]

    return root

# Merge Tree Based on Rules
def merge_tree(x, y):

    # Merged tree root - take Y's value
    merged = Node(x.key, y.value)

    # Isolate Y's children
    y_children = {child.key: child for child in y.children}

    # Track key's visited
    merged_keys = set() 

    # X first to preserve ordering
    for x_child in x.children:

        if x_child.key in y_children:

            # Exists in both
            y_child = y_children[x_child.key]
            # Recurvisvely add
            merged.children.append(merge_tree(x_child, y_child))

        else:
            # Only in X, take whole subtree
            merged.children.append(x_child) 

        merged_keys.add(x_child.key)

    # Add unique children of Y
    for y_child in y_children:

        if y_child.key not in merged_keys:
            merged.children.append(y_child) 

    return merged


# Main Function
def solution(arg1, arg2):

    # Use previous code to create the trees
    left_tree = create_tree(arg1) 
    right_tree = create_tree(arg2) 

    # Step 1 - different keys 
    if left_tree.key != right_tree.key:
        return []

    # Step 2 - merge trees
    new_tree = merge_tree(left_tree, right_tree)

    # Traverse tree and return result
    result = preorder(new_tree)

    return result
```
