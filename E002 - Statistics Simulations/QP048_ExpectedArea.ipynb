Two integers X and Y are chosen independently and uniformly from the set {1,2,3,4,5}. Note that since the selection is independent, X and Y may be equal. These values serve as the coordinates of a right triangle with vertices at 
(0,0), (X,0), and (0,Y). What is the expected area of this triangle?

Ans = 4.5

```python
import random as rand

X = [1, 2, 3, 4, 5]
Y = [1, 2, 3, 4, 5] 
N = 1000000
A = []
for i in range(N):
    x = rand.choice(X) 
    y = rand.choice(Y) 
    area = 0.5*x*y 
    A.append(area) 

print(sum(A)/N)
```
