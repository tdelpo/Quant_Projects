Three ants are positioned on the three vertices of an equilateral triangle. Each ant randomly chooses one of the two adjacent vertices to move to, with equal probability. What is the probability that two ants will meet at the same vertex?

Ans = 0.75

```python
import random as rand

ANT = ["L", "R"] 
N = 1000000
count = 0
for _ in range(N):
    a1 = rand.choice(ANT)
    a2 = rand.choice(ANT)
    a3 = rand.choice(ANT)
    combine = a1+a2+a3 
    if combine == "LLL" or combine == "RRR":
        count += 1

print(1 - count/N)
```
