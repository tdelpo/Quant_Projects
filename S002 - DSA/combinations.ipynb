```python
import itertools as iter 
```

```python
# Functions 
def method_perms(s): 

    perms = set("".join(p) for p in iter.permutations(s, len(s)))

    return perms 

def recursive_perms(s):

    from collections import Counter 

    counts = Counter(s) 
    result = [] 
    temp = [] 
    n = len(s) 

    def rloop_perms(): 
        if len(temp) == n:
            result.append("".join(temp)) 
            return 
        
        for char in counts:
            if counts[char] > 0: 
                temp.append(char) 
                counts[char] -= 1 

                rloop_perms() 

                counts[char] += 1
                temp.pop() 
    
    rloop_perms() 

    return result
```

```python
s = 'hello' 
perms_1 = method_perms(s)
print(len(perms_1)) 
print(sorted(perms_1))
```

```python
perms_2 = recursive_perms(s)
print(len(perms_2)) 
print(sorted(perms_2))
```
