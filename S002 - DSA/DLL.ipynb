```python
class Node: 

    "Class to create nodes and their inherent characteristics."

    def __init__(self, value): 
        self.value = value 
        self.next = None 
        self.prev = None
```

```python
class DoublyLinkedList: 

    "Class to handle doubly linked lists and their methods." 

    def __init__(self, value):
        "Initialise the DLL class with first node."
        new_node = Node(value)
        self.head = new_node 
        self.tail = new_node 
        self.length = 1 

    def print_list(self):
        "Print current list of values."
        temp = self.head 
        while temp is not None:
            print(temp.value) 
            temp = temp.next

    def append(self, value):
        "Append a new node to the dll."
        new_node = Node(value)
        if (self.head is None):
            self.head = new_node 
            self.tail = new_node     
        else:
            self.tail.next = new_node 
            new_node.prev = self.tail 
            self.tail = new_node
        self.length += 1
        return True 
    
    def pop(self): 
        "Remove last node from dll." 
        if self.length == 0:
            return None
        temp = self.tail 
        if self.length == 1:
            self.head = None 
            self.tail = None 
        else: 
            self.tail = self.tail.prev 
            self.tail.next = None 
            temp.prev = None 
        self.length -= 1 
        return temp
    
    def prepend(self, value):
        "Add new node to beginning of dll." 
        new_node = Node(value) 
        if self.length == 0:
            self.head = new_node 
            self.tail = new_node 
        else:
            new_node.next = self.head 
            self.head.prev = new_node 
            self.head = new_node
        self.length += 1
        return True
    
    def pop_first(self):
        "Remove first node of dll." 
        if self.length == 0:
            return None 
        temp = self.head 
        if self.length == 1:
            self.head = None 
            self.tail = None 
        else:
            self.head = self.head.next 
            self.head.prev = None 
            temp.next = None 
        self.length -= 1 
        return temp 
    
    def get(self, index): 
        "Return node at given index." 
        if index < 0 or index >= self.length:
            return None 
        temp = self.head
        if index < self.length/2:
            for _ in range(index):
                temp = temp.next
        else:
            temp = self.tail   
            for _ in range(self.length -1, index, -1):
                temp = temp.prev 
        return temp 
    
    def set_value(self, index, value):  
        "Set the value of a given node."
        temp = self.get(index) 
        if temp:
            temp.value = value 
            return True 
        return False 
    
    def insert(self, index, value): 
        "Insert new node at given index." 

        if index < 0 or index > self.length:
            return False 
        if index == 0:
            return self.prepend(value) 
        if index == self.length:
            return self.append(value)
              
        new_node = Node(value)
        before = self.get(index - 1) 
        after = before.next 

        new_node.prev = before 
        new_node.next = after
        before.next = new_node 
        after.prev = new_node 

        self.length += 1
        return True
    
    def remove(self, index):

        if index < 0 or index > self.length:
            return None 
        if index == 0:
            return self.pop_first() 
        if index == self.length:
            return self.pop() 
        temp = self.get(index) 
        temp.prev.next = temp.next 
        temp.next.prev = temp.prev 
        temp.prev = None 
        temp.next = None 

        self.length -= 1 
        return temp


```
