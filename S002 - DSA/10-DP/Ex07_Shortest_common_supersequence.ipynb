Shortest common supersequence (problem)
Given two strings s1 and s2, find the length of their shortest common supersequence, the shortest string that has both s1 and s2 subsequences.

A string A is a supersequence of a string B if B is a subsequence of A.

Example:

input:
s1 = "abdacebfcab"
s2 = "acebfca"

output: 11

explanation: The shortest common supersequence of s1 and s2 is "abdacebfcab", its length is 11

```python
# Function
def scs(s1, s2):

    n = len(s1)
    m = len(s2) 
    dp = [[0]*(m+1) for _ in range(n+1)]

    # Base cases
    for i in range(1,n+1):
        dp[i][0] = i 
    for j in range(1,m+1):
        dp[0][j] = j 
        

    # Loop
    for i in range(1, n+1):
        for j in range(1, m+1):
            if s1[i-1] == s2[j-1]:
                dp[i][j] = dp[i-1][j-1] + 1
            else:
                dp[i][j] = 1 + min(dp[i-1][j], dp[i][j-1])

    return dp[-1][-1]

# Recursive
def scs_r(s1, s2, i=0, j=0, lookup=None):
    lookup = {} if lookup is None else lookup
    # Base cases
    if (i,j) in lookup:
        return lookup[(i,j)]
    if i == len(s1):
        return len(s2[j:])
    if j == len(s2):
        return len(s1[i:])
    elif s1[i] == s2[j]:
        lookup[(i,j)] = 1 + scs_r(s1, s2, i+1, j+1, lookup)
        return lookup[(i,j)]
    else:
        lookup[(i,j)] = 1 + min(scs_r(s1, s2, i+1, j, lookup), scs_r(s1, s2, i, j+1, lookup))
        return lookup[(i,j)] 
    
```

```python
# Test case
s1 = "abdacebfcab"
s2 = "acebfca"

# Run
print(scs(s1, s2))

# Run
print(scs_r(s1, s2))
```
