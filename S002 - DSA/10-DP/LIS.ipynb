Longest increasing subsequence (problem)
Given an array of integers arr, find the length of its longest increasing subsequence, its longest subsequence where each element is strictly greater than the previous one.



Example 1:

input:
arr = [7, 5, 2, 4, 7, 2, 3, 6, 4, 5, 12, 1, 7]

output: 5

explanation: A possible longest increasing subsequence is [2, 3, 4, 5, 7], its length is 5



Example 2:

input:
arr = [8, 5, 5, 3]

output: 1

explanation: The longest increasing subsequences that we can make are those that contain one element only, like [8]



Constraints:

len(arr) >= 1

```python
from bisect import bisect_left

# Function
def lis(arr):

    if len(arr) == 0:
        return 0

    mem = []

    for num in arr:

        idx = bisect_left(mem, num)
        if idx == len(mem):
            mem.append(num) 
        else:
            mem[idx] = num
   
    return len(mem)

# Function
from math import inf
def LIS(arr, i=0, prev=-inf, lookup=None):

    lookup = {} if lookup is None else lookup 
    if (i, prev) in lookup:
        return lookup[(i, prev)]
    if i == len(arr):
        return 0
    else:
        lookup[(i, prev)] = max(
            (1 + LIS(arr, i+1, prev=arr[i]) if arr[i]>prev else 0),
            LIS(arr, i+1, prev=prev)
        )
        return lookup[(i, prev)]

```

```python
# Test Case 1
arr = [7, 5, 2, 4, 7, 2, 3, 6, 4, 5, 12, 1, 7]

# Run
print(LIS(arr))

# Test Case 2
arr = [8, 5, 5, 3]

# Run
print(LIS(arr))
```
