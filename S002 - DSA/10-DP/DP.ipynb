```python
def fib(n):
    global counter 
    counter += 1
    if memo[n] is not None:
        return memo[n]
    if n == 0 or n == 1:
        return n
    memo[n] = fib(n-1) + fib(n-2)
    return memo[n]
```

```python
memo = [None] * 100
counter = 0 
n = 35 

print('Fib of', n, '=', fib(n)) 
print('Counter:', counter)
```

```python
def fib_2(n):
    global counter_2 
    fib_list = [0, 1] 
    for index in range(2, n+1): 
        counter_2 += 1
        next_fib = fib_list[index-1] + fib_list[index-2] 
        fib_list.append(next_fib)
    return fib_list[n] 
```

```python
counter_2 = 0 
n = 35 

print('Fib of', n, '=', fib_2(n)) 
print('Counter:', counter_2) 
```
