Partition (problem)
Given an array of strictly positive integers arr, check if we can split it into two subsets that have the same sum of elements.



Example 1:

input:
arr = [4, 5, 3, 2, 5, 1]

output: true

explanation: We can split arr into [4, 3, 2, 1] and [5, 5], and they have the same sum
4+3+2+1 = 5+5 = 10



Example 2:

input:
arr = [5, 6, 2, 3, 8, 1]

output: false

explanation: We can't split arr into two subsets that have the same sum



Constraints:

len(arr) >= 1
arr[i] >= 1

```python
# Function
def subsets(arr, k, i=0, lookup=None):
    lookup = {} if lookup is None else lookup
    if (i, k) in lookup:
        return lookup[(i,k)]
    if k == 0:
        return 1
    elif i == len(arr) or k < 0:
        return 0
    else:
        lookup[(i,k)] = subsets(arr, k-arr[i], i+1) + subsets(arr, k, i+1)
        return lookup[(i,k)]  
```

```python
# Function - recursive
def partition(arr, i=0, sum1=0, sum2=0):

    if i == len(arr):
        return sum1 == sum2
    else:
        return partition(arr, i+1, sum1+arr[i], sum2) or partition(arr, i+1, sum1, sum2+arr[i])

def partition(arr):
    s = sum(arr)
    if s%2 == 1:
        return False
    else:
        return subsets(arr, s) > 0
```

```python
# Test 
arr = [4, 5, 3, 2, 5, 1]
print(partition(arr))

# Test 
arr = [5, 6, 2, 3, 8, 1]
print(partition(arr))
```
