Gold mine (problem)
Given a mine of n rows and m columns where mine[i][j] represents the amount of gold that is present there, we want to enter from the top of the mine and take as much gold as possible when exiting from the bottom, knowing that we can only move to the bottom, to the bottom-left, or to the bottom-right. We can exit from anywhere from the last row.

Example:

input:
mine = [
   [3, 2, 12, 15, 10],
   [6, 19, 7, 11, 17],
   [8, 5, 12, 32, 21],
   [3, 20, 2, 9, 7]
]

output: 73

explanation: 15+17+32+9 = 73

Constraints:
len(matrix) >= 1
len(matrix[i]) >= 1
matrix[i][j] >= 0

```python
# Test Case
mine = [
   [3, 2, 12, 15, 10],
   [6, 19, 7, 11, 17],
   [8, 5, 12, 32, 21],
   [3, 20, 2, 9, 7]
]

# Edge case - single row
mine_1 = [[3, 2, 12, 15, 10]] 

# Edge case - single column
mine_2 = [
   [3],
   [6],
   [8],
   [3]
]
```

```python
def gold(mine):

    n = len(mine)
    m = len(mine[0]) 
    print(n, m)

    # Edge case 2 - single column
    if m == 1:
        ans = 0
        for i in range(n):
            ans += mine[i][0]
        return ans

    # DP Matrix
    dp = [[0]*m for _ in range(n)] 

    # First row
    dp[0] = mine[0]

    for i in range(1,n):
        for j in range(m):
            if j == 0:
                dp[i][j] = mine[i][j] + max(dp[i-1][j], dp[i-1][j+1])
            elif j == (m-1):
                dp[i][j] = mine[i][j] + max(dp[i-1][j-1], dp[i-1][j])
            else:
                dp[i][j] = mine[i][j] + max(dp[i-1][j-1], dp[i-1][j], dp[i-1][j+1])

    return max(dp[-1])
```

```python
# Main
answer = gold(mine)
print(answer)

# Edge 1
answer = gold(mine_1)
print(answer)

# Edge 2
answer = gold(mine_2)
print(answer)
```
