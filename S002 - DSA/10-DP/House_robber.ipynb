```python
def rob_r(arr, i=0, lookup=None):

    # Memo
    lookup = {} if lookup is None else lookup
    if i in lookup:
        return lookup[i]
    
    # Edge case - single house
    if len(arr) == 1:
        return arr[0] 
    
    n = len(arr) 
    if i == n:
        return 0
    if i == n-1:
        lookup[i] = arr[i]
        return lookup[i]
    else:
        lookup[i] = max(arr[i] + rob_r(arr, i+2, lookup), rob_r(arr,i+1, lookup))
        return lookup[i]

def rob_t(arr):

    n = len(arr)
    if n == 1:
        return arr[0]
    
    dp = [0 for _ in range(n)]
    dp[0] = arr[0]
    dp[1] = max(arr[0], arr[1])

    if n > 2:
        for i in range(2,n):
            dp[i] = max(arr[i] + dp[i-2], dp[i-1])

    return dp[-1]
```

```python
arr = [2, 10, 3, 6, 8, 1, 7]

# Recursive
print(rob_r(arr))

# Tabulation
print(rob_t(arr))
```
